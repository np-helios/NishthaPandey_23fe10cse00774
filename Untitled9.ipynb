```python
def cap_outliers(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return column.clip(lower, upper)

X_capped = X.copy()
for col in X_capped.columns:
    X_capped[col] = cap_outliers(X_capped[col])
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

wcss = []

for i in range(1, 10):
    km = KMeans(n_clusters=i, random_state=42)
    km.fit(X_scaled)
    wcss.append(km.inertia_)

plt.plot(range(1, 10), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
k = 3

kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(X_scaled)

df['Cluster'] = kmeans.labels_
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df['Cluster'])
plt.title("Country Clusters (K-Means)")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.show()
```

```python
print("\nCluster-wise Mean Values:")
print(df.groupby('Cluster').mean(numeric_only=True))

# Top countries in each cluster
for i in range(k):
    print(f"\nCountries in Cluster {i}:")
    print(df[df['Cluster'] == i]['country'].values)
```
